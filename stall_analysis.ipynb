## **Introduction**
This notebook analyses two weeks of sales and stock records for Wanjiku's fresh-produce stall at Gikomba Market.

The analysis uses core Python to evaluate the sales performance, identify business trends, classify daily sales and determine which products require restocking.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2


# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: Sales Performance

I will use Python's built-in functions to calculate the total, average, highest and lowest sales over the 14-day period. I will also use list indexing to identify the corresponding day and week.

```python
# Calculate sales statistics
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

# Identify the positions of the highest and lowest sales
best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

# Identify day names
best_day = days[best_index]
worst_day = days[worst_index]

# Identify week numbers
best_week = (best_index // 7) + 1
worst_week = (worst_index // 7) + 1

# Display results
print("GIKOMBA STALL: SALES OVERVIEW")
print("-" * 30)

print(f"Total sales: KES {total_sales:,.2f}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest sales: KES {highest_sales:,.2f}")
print(f"Best day: {best_day}, Week {best_week}")
print(f"Lowest sales: KES {lowest_sales:,.2f}")
print(f"Worst day: {worst_day}, Week {worst_week}")
```

## Task 2: Weekly Sales Comparison and Growth

I will use list slicing to separate the 14 days into two weeks. A reusable function will calculate the total sales for each week and return the results as a tuple.

I wil then calculate the percentage change between week 1 and week 2.


```python
def week_totals(sales):
    """
    Calculate and return the sales totals for two weeks.
    Returns:
        A tuple containing Week 1 and Week 2 totals.
    """

    week1_total = sum(sales[:7])
    week2_total = sum(sales[7:])

    return week1_total, week2_total

week1, week2 = week_totals(daily_sales)

# Calculate each week's percentage contribution
week1_percentage = (week1/total_sales)*100
week2_percentage = (week2/total_sales)*100

# Calculate week-on-week growth percentage
growth_percentage = ((week2 - week1) / week1) * 100

# Display results
print("\nWEEKLY SALES PERFORMANCE")
print("------------------------")
print(f"Week 1 sales: KES {week1:,.2f}")
print(f"Week 2 sales: KES {week2:,.2f}")
print(f"Week 1 contribution: {week1_percentage:.2f}%")
print(f"Week 2 contribution: {week2_percentage:.2f}%")
print(f"Week-on-week growth: {growth_percentage:.2f}%")
```

## Task 3: Daily Sales Classification

I will create a function called 'classify_day()' to classify daily sales based on
the overall average. Days with sales more than 20% above
the average will be classified as ***Strong***, while days more than 20% below the average will be classified as ***Weak***.
All other days within the remaining range will be classified as ***Normal***.

I will then apply the function to all 14 days and use a dictionary to count the number of days in each category.

```python
print("Sales Category Summary")
print("----------------------")
def classify_day(amount, average):
    """
    Classify daily sales as Strong, Normal, or Weak.
    Returns:
      A string representing the sales category
    """

    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"

daily_categories = []
for sale in daily_sales:
    category = classify_day(sale, average_sales)
    daily_categories.append(category)

print(daily_categories)

category_counts = {}
for category in daily_categories:
    category_counts[category] = category_counts.get(category, 0) + 1

for category, count in category_counts.items():
    print(f"{category}: {count} days")
```

## Task 4: Weekly vs Weekend Sales Analysis

I will separate daily sales into weekdays (Monday to Friday)
and weekends (Saturday and Sunday).

I will calculate the average sales for each group and
determine the percentage difference, using weekday average sales as the baseline.

```python
# Separating weekday and weekend sales
weekday_sales = []
weekend_sales = []

for i in range(len(daily_sales)):
    if days[i] in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
        weekday_sales.append(daily_sales[i])
    else:
        weekend_sales.append(daily_sales[i])

# Calculate averages
weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

# Calculate the percentage difference
percentage_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

# Results
print("Weekly vs Weekend Sales Analysis")
print("--------------------------------")
print(f"Weekday average sales: KES {weekday_average:,.2f}")
print(f"Weekend average sales: KES {weekend_average:,.2f}")
print(f"Weekend sales are {percentage_difference:.2f}% higher than weekday sales.")
```

## Task 5: Inventory Management and Gross Margins

I will examine the stock dictionary to identify products
whose current quantities are at or below their reorder
levels.

I will also calculate the gross profit margin for each
product using its cost and selling price.

```python
def products_to_reorder(stock):
    """
    Identify products that have reached or fallen below
    their reorder levels.

    Returns:
        A list of product names that need restocking.
    """

    reorder_list = []

    for product, details in stock.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list

reorder_products = products_to_reorder(stock)

print("Products Requiring Restocking")
print("-----------------------------")

for product in reorder_products:
    print(product)

# Calculate gross profit margins
print("\nGross Profit Margins")
print("--------------------")

for product, details in stock.items():
    margin = (
        (details["price"] - details["cost"])
        / details["price"]
    ) * 100

    print(f"{product}: {margin:.2f}%")
```

## Task 6: Stall Performance Report

I will combine the sales, weekly growth, daily classifications,
weekday and weekend comparisons, inventory alerts and gross profit margins into a formatted report using Python f-strings.

```python
print("\n" + "=" * 40)
print("       STALL PERFORMANCE REPORT")
print("=" * 40)

# Overall sales performance
print("\n1. OVERALL SALES PERFORMANCE")
print(f"Total sales: KES {total_sales:,.2f}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest daily sales: KES {max(daily_sales):,.2f}")
print(f"Lowest daily sales: KES {min(daily_sales):,.2f}")
print(f"Best sales day: {best_day}, Week {best_week}")
print(f"Worst sales day: {worst_day}, Week {worst_week}")

# Weekly performance
print("\n2. WEEKLY SALES PERFORMANCE")
print(f"Week 1: KES {week1:,.2f} ({week1_percentage:.2f}%)")
print(f"Week 2: KES {week2:,.2f} ({week2_percentage:.2f}%)")
print(f"Week-on-week growth: {growth_percentage:.2f}%")

# Daily sales classification
print("\n3. DAILY SALES CLASSIFICATION")
for category, count in category_counts.items():
    print(f"{category}: {count} days")

# Weekday vs weekend
print("\n4. WEEKDAY VS WEEKEND PERFORMANCE")
print(f"Weekday average: KES {weekday_average:,.2f}")
print(f"Weekend average: KES {weekend_average:,.2f}")
print(f"Weekend increase: {percentage_difference:.2f}%")

# Inventory alerts
print("\n5. INVENTORY ALERTS")
for product in reorder_products:
    print(f"Restock: {product}")

# Gross profit margins
print("\n6. GROSS PROFIT MARGINS")
for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100
    print(f"{product}: {margin:.2f}%")

print("\n" + "=" * 40)
print("             END OF REPORT")
print("=" * 40)
```
